## Basic API Request

Now, let's  make requests to the OCD API. We'll use a sample image to test our function.

```python
# let's check the image first
image_path = 'check.png'
image = plt.imread(image_path) # read image file
plt.imshow(image) #Prepare to display this image
plt.show()
```

```python
endpoint = "https://api.ocr.space/parse/image" # this is destination URL

```

```python
# Open the image file
with open(image_path, "rb") as image_file: #open image as read binary
    # Set up the request payload
    payload = {
        "apikey": api_key # this data payload is where u put all the non file information the API need
    }

    # Send the request
    response = requests.post(endpoint, files={"file": image_file}, data=payload) #send file to server to the address of endpoint

# Parse the response
result = response.json()
```

```python
# Parse the response
result = response.json()

# Check if the API returned a valid response
if "ParsedResults" in result:
    for parsed_result in result["ParsedResults"]:
        print("Extracted Text:", parsed_result["ParsedText"])
        print("\nBounding Boxes:")

else:
    print("Error:", result.get("ErrorMessage", "Something went wrong"))
```
