```python
import cv2
```

```python
import numpy as np
```

```python
from matplotlib import pyplot as plt
```

```python
plt.rcParams['figure.figsize'] = [10,5]
```

```python
img = cv2.imread('img.png')
```

```python
plt.imshow(img)
```

```python
img.shape
```

```python
rows, cols, channels = img.shape
```

```python
img_rgb = np.zeros_like(img)
```

```python
for r in range(rows):
  for c in range(cols):
    pixel = img[r,c, :]
    img_rgb[r,c,0] = pixel[2]
    img_rgb[r,c,1] = pixel[1]
    img_rgb[r,c,2] = pixel[0]
plt.imshow(img_rgb)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img_rgb)
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap = 'gray')
```

```python
img_gray
```

```python
blue, green, red = cv2.split(img)
```

```python
img_blue = np.zeros_like(img)

for r in range(rows):
  for c in range(cols):
    img_blue[r,c,0] = 0
    img_blue[r,c,1] = 0
    img_blue[r,c,2] = blue[r,c]
```

```python
plt.imshow(img_blue)
```
